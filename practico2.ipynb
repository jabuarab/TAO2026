```python
import cvxpy as cp
import numpy as np

c = np.array([
    [5, 4, 1, 3],
    [3, 1, 4, 2],
    [4, 2, 3, 1]
])

s = np.array([110, 100, 100])
d = np.array([30, 60, 90, 120])

n, m = c.shape

x = cp.Variable((n, m))


h = cp.sum(x, axis=0) == d
g1 = cp.sum(x, axis=1) <= s
g2=  -x <= 0

obj = cp.Minimize(cp.sum(cp.multiply(c, x)))

problem = cp.Problem(obj, [h, g1,g2])
problem.solve()

print("Costo óptimo:", problem.value)
print("Cantidades óptimas X:")
print(np.round(x.value, 4))

print("mu:")
print(g1.dual_value)
print("rrho:")
print(g2.dual_value)

print("lambda:")
print(h.dual_value)
```
