```python
import cvxpy as cp
import numpy as np

c = np.array([
    [5, 4, 1, 3],
    [3, 1, 4, 2],
    [4, 2, 3, 1]
])

s = np.array([110, 100, 100])
d = np.array([30, 60, 90, 120])

n, m = c.shape

x = cp.Variable((n, m))


h = cp.sum(x, axis=0) == d
g1 = cp.sum(x, axis=1) <= s
g2=  -x <= 0

obj = cp.Minimize(cp.sum(cp.multiply(c, x)))

problem = cp.Problem(obj, [h, g1,g2])
problem.solve()

print("Costo óptimo:", problem.value)
print("Cantidades óptimas X:")
print(np.round(x.value, 4))

print("mu:")
print(g1.dual_value)
print("rrho:")
print(g2.dual_value)

print("lambda:")
print(h.dual_value)
```

```python
casos=[]
z = np.array([1.2, 0.5, -0.1, 0.2])
x = np.array([0,0,0,0])
mu = np.array([0,0,0,0])
for i in range(16):
    caso = [int(bit) for bit in f"{i:04b}"]
    casos.append(caso)
for caso in casos:
    mu_no_activos = np.array(caso) == 0
    mu_activos = np.array(caso) == 1

    #dL/dxi=0
    #x_i = z_i + lam +  0
    #sum  x  = 1
    #sum  z_i + lam  =1  
    #k lam=  1 -  sum z_i
    #k=  #x=/=0!!! no  es  N

    k=np.sum(mu_no_activos)
    lam = (1-np.sum(z[mu_no_activos]) ) / k

    x = np.zeros(4)
    mu = np.zeros(4)

    x[mu_no_activos] = z[mu_no_activos] + lam

    #x=0
    mu[mu_activos] = -(lam + z[mu_activos])

    # Comprobar factibilidad primal y dual
    cumple = (
        np.all(x >= 0)
        and np.all(mu >= 0)
        and (np.sum(np.round(x, 4)) - 1) == 0
    )

    print(
        caso,
        "lam:", round(lam, 4),
        "x :", np.round(x, 4),
        "mu :", np.round(mu, 4),
        "f",0.5*(np.mean(x-z)**2),
        cumple
    )
```
